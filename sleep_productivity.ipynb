feature engineering

```python
condition= [df['Productivity Score'] < 4,
             df['Productivity Score'] < 7,
             df['Productivity Score'] >= 7]

categories= ['Low', 'Medium', 'High']

df['Productivity_Category']= np.select(condition, categories,default='Unknown')
```

```python
df[['Productivity Score','Productivity_Category']].head(10)
```

```python
condition=[df['Mood Score'] < 4,
           df['Mood Score'] <= 7,
           df['Mood Score'] > 7] 
categories=['Low','Medium','High']
df['Mood_Category']=np.select(condition,categories,default='Unknown')
```

```python
df[['Mood Score','Mood_Category']].head(10)
```

```python
condition=[df['Caffeine Intake (mg)'] < 100,
           df['Caffeine Intake (mg)'] < 200]
categories=('Low','Medium')
df['Caffeine_Category']=np.select(condition,categories,default='High')
```

```python
df[['Caffeine Intake (mg)','Caffeine_Category']].head(10)
```
